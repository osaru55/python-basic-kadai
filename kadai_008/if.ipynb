```python
"""
・変数varが、3の倍数の場合は「Fizz」を出力
・変数varが、5の倍数の場合は「Buzz」を出力
・変数varが、3の倍数と5の倍数の両方に該当する場合は「FizzBuzz」を出力
・上記のどの場合にも該当しない場合は、変数varの値を出力

ただし、変数varは正の整数とします。
"""

# 変数varに正の整数を代入する
var = 15

# 変数varの値を出力
print(var)

# 変数answerの値をクリア
answer = ""

#3の倍数の時
if var % 3 == 0:
    answer = "Fizz"
#5の倍数の時
if var % 5 == 0:
    answer += "Buzz"

#3または5の倍数のいずれでもないとき
if answer == "":
  print(var)
#3または5の倍数のとき
else:
  print(answer)


```
